```python
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    Image
)

from reportlab.lib.styles import getSampleStyleSheet
```

```python
pdf_path = "../reports/Apple_Financial_Analysis_Report.pdf"
```

```python
doc = SimpleDocTemplate(
    pdf_path
)
```

```python
story = []
```

```python
styles = getSampleStyleSheet()

title = Paragraph(
    "Apple Financial Analysis Report",
    styles["Title"]
)

story.append(title)

story.append(Spacer(1,20))
```

```python
summary = """
Executive Summary

Apple demonstrated stable revenue growth from 2021 to 2024,
with revenue increasing by 6.89%.

However, net income declined during the period,
indicating potential profitability pressure.

The company maintained a strong gross margin of 43.86%,
reflecting strong pricing power and operational efficiency.

Apple's debt ratio reached 84.40%, showing a relatively
high leverage structure, although strong cash generation
helps mitigate financial risk.
"""


story.append(
    Paragraph(
        summary,
        styles["BodyText"]
    )
)

story.append(
    Spacer(1,20)
)
```

```python
story.append(
    Image(
        "../images/revenue_trend.png",
        width=400,
        height=250
    )
)

story.append(
    Spacer(1,20)
)
story.append(
    Image(
        "../images/gross_margin.png",
        width=400,
        height=250
    )
)

story.append(
    Spacer(1,20)
)
story.append(
    Image(
        "../images/net_income.png",
        width=400,
        height=250
    )
)

story.append(
    Spacer(1,20)
)
story.append(
    Image(
        "../images/free_cash_flow.png",
        width=400,
        height=250
    )
)

story.append(
    Spacer(1,20)
)
```

```python
doc.build(story)
```

```python
revenue_growth = (
    income["Revenue"].iloc[-1]
    -
    income["Revenue"].iloc[0]
) / income["Revenue"].iloc[0] * 100


revenue_growth
```

```python
profit_growth = (
    income["Net_Income"].iloc[-1]
    -
    income["Net_Income"].iloc[0]
) / income["Net_Income"].iloc[0] * 100


profit_growth
```

```python
avg_gross_margin = (
    income["Gross_Margin"].mean()
)


avg_gross_margin
```

```python
latest_debt_ratio = (
    balance["Debt_Ratio"].iloc[-1]
)


latest_debt_ratio
```

```python
insights = []
```

```python
if revenue_growth > 0:

    insights.append(
        f"Revenue increased by {revenue_growth:.2f}% "
        "from 2021 to 2024, indicating stable business growth."
    )

else:

    insights.append(
        "Revenue declined during the analysis period."
    )
if profit_growth > 0:

    insights.append(
        "Net income improved, showing stronger profitability."
    )

else:

    insights.append(
        "Net income decreased, indicating potential margin pressure."
    )
insights.append(
    f"Average gross margin was {avg_gross_margin:.2f}%, "
    "reflecting Apple's pricing power and operational efficiency."
)
if latest_debt_ratio > 80:

    insights.append(
        f"Debt ratio reached {latest_debt_ratio:.2f}%, "
        "indicating a highly leveraged capital structure."
    )

else:

    insights.append(
        "Debt level remains moderate."
    )
```
